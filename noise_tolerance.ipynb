```python
import cv2
import seaborn as sns
import matplotlib.pyplot as plt
import skimage as ski
import numpy as np
```

```python
filename = 'test_images/0001.png'
mean = 0
var = 1

image = ski.io.imread(filename)
gray_img = ski.color.rgb2gray(image) 
noisy_img_float = ski.util.random_noise(gray_img, mode='speckle', mean=mean, var=var)
rgb_noisy_img = ski.util.img_as_ubyte(ski.color.gray2rgb(noisy_img_float)) # Convert the image to unint8 and rgb
```

```python
# Visualization
plt.imshow(image)
plt.show()
plt.imshow(rgb_noisy_img)
plt.show()
```

```python
# Save the image
out_filename = 'noisy_images/0001_speckle_m{}v{}.png'.format(mean, var)
ski.io.imsave(out_filename, rgb_noisy_img)
```
