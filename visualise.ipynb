```python
import cv2
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
or_filename="test_images/0000.png"
original = cv2.imread(or_filename)
```

```python
filename="test_outputs/0000_out.png"
image = cv2.imread(filename)
```

```python
gray_img = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Original image
plt.subplot(131)
plt.imshow(original)
plt.title("original")

# the image with masks
plt.subplot(132)
sns.heatmap(gray_img, xticklabels=False, yticklabels=False,alpha=0.5, cbar=False, cmap="mako")
plt.imshow(original)
plt.title("predicted masks")

# # masks
# plt.subplot(133)
# sns.heatmap(gray_img, xticklabels=False, yticklabels=False, cbar=False, cmap="mako")

plt.tight_layout()
plt.show()
```
