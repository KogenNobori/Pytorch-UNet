# Check if masks and images correspond

```python
import skimage as ski
import matplotlib.pyplot as plt
```

```python
example = ski.io.imread('data/imgs/0500.png')
example_mask = ski.io.imread('data/masks/0500.png')

plt.subplot(211)
plt.imshow(example)

plt.subplot(212)
plt.imshow(example_mask*50)
plt.title('mask')

plt.tight_layout()
plt.show()
```
